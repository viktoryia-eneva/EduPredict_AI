### Academic performance by outcome

Students who graduate tend to pass more curricular units in both the first and second semesters.

- Dropout students pass around 2-3 units on average.
- Enrolled students pass around 4 units.
- Graduate students pass around 6 units.

This suggets that early academic performance may be an important predictor of student outcome.

```python
df["Dropout_Risk"] = df["Target"].apply(lambda x: 1 if x == "Dropout" else 0)
```

```python
df["Dropout_Risk"].value_counts()
```

```python
early_features = [
    "Application mode",
    "Application order",
    "Course",
    "Daytime/evening attendance\t",
    "Previous qualification",
    "Previous qualification (grade)",
    "Admission grade",
    "Displaced",
    "Educational special needs",
    "Debtor",
    "Tuition fees up to date",
    "Gender",
    "Scholarship holder",
    "Age at enrollment",
    "International",
    "Curricular units 1st sem (credited)",
    "Curricular units 1st sem (enrolled)",
    "Curricular units 1st sem (evaluations)",
    "Curricular units 1st sem (approved)",
    "Curricular units 1st sem (grade)",
    "Curricular units 1st sem (without evaluations)"
]

X_early = df[early_features]
y = df["Dropout_Risk"]
```

```python
X_early.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_early,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
y_pred[:20]
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

rf_accuracy = accuracy_score(y_test, rf_pred)
rf_accuracy
```

```python
print(classification_report(y_test, rf_pred))
```

```python
confusion_matrix(y_test, rf_pred)
```

```python
import pandas as pd

feature_importance = pd.Series(
    rf_model.feature_importances_,
    index=X_train.columns
)

feature_importance.sort_values(ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
top_features = feature_importance.sort_values(ascending=False).head(10)

top_features.sort_values().plot(kind="barh")

plt.title("Top 10 Features for Early Dropout Prediction")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.show()
```

### Key finding

First-semester academic performance was the strongest source of predictive information in the Random Forest model. The number of approved curricular units and first-semester grades were more important than admission grade or demographic characteristics.

Financial and enrollment-related factors, such as tuition fee status and age at enrollment, also contributed to dropout-risk prediction.

Feature importance describes predictive usefulness and should not be interpreted as evidence of causation.

```python
X_full = df.drop(columns=["Target", "Dropout_Risk"])

X_full_train = X_full.loc[X_train.index]
X_full_test = X_full.loc[X_test.index]
```

```python
X_full_train.shape, X_full_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_full = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf_full.fit(X_full_train, y_train)
```

```python
rf_full_pred = rf_full.predict(X_full_test)
```

```python
full_accuracy = accuracy_score(y_test, rf_full_pred)

full_accuracy
```

```python
print(classification_report(y_test, rf_full_pred))
```

```python
confusion_matrix(y_test, rf_full_pred)
```

```python
import pandas as pd
from sklearn.metrics import (
    precision_score, 
    recall_score, 
    accuracy_score, 
    f1_score
    )

comparison = pd.DataFrame({
    "Model": [
        "Early Random Forest",
        "Full Random Forest"
    ],
    "Accuracy" : [
        accuracy_score(y_test, rf_pred),
        accuracy_score(y_test, rf_full_pred)
    ],
    "Precision (Dropout)": [
        precision_score(y_test, rf_pred),
        precision_score(y_test, rf_full_pred)
    ],
    "Recall (Dropout)": [
        recall_score(y_test, rf_pred),
        recall_score(y_test, rf_full_pred)
    ],
    "F1-score (Dropout)": [
        f1_score(y_test, rf_pred),
        f1_score(y_test, rf_full_pred)
    ]       
})

comparison
```

```python
comparison_plot = comparison.set_index("Model")

comparison_plot[[
    "Accuracy",
    "Precision (Dropout)",
    "Recall (Dropout)",
    "F1-score (Dropout)"
]].plot(kind="bar", figsize=(10,6))

plt.title("Early vs Full Random Forest Performance")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc="lower right")

plt.show()
```

### Model Comparison

The Full Random Forest model achieved the best overall performance.

- Early Random Forest accuracy: approximately 86.2%
- Full Random Forest accuracy: approximately 88.2%
- Early dropout recall: approximately 69.0%
- Full dropout recall: approximately 74.3%

Including second-semester information improves dropout detection. However, the Early Model still provides relatively strong performance while allowing at-risk students to be identified sooner.
This demonstrates an important trade-off between predictive performanceand early intervation.

### Limitations

This project uses data from a single higher-education institution, so the results may not generalize directly to students from other universities or countries.

The dataset also contains historical student information. Therefore, the model should be treated as a proof-of-concept rather than a system ready for real-world decision-making.

Feature importance shows which variables were useful for prediction, but it does not prove that those variables cause student dropout.

The model should be validated on new and more diverse datasets before being used in practice.

### Final Conclusion

This project explored whether machine learning can be used to identify students at risk of dropping out.

Two Random Forest models were compared:

- An Early Model using enrollment and first-semester information.
- A Full Model including second-semester information.

The Early Random Forest achieved approximately 86.2% accuracy and 69.0% recall for dropout students.

The Full Random Forest improved performance to approximately 88.2% accuracy and 74.3% dropout recall.

Although the Full Model performed better, the Early Model still produced relatively strong predictions while identifying risk earlier.

```python
def get_risk_level(probability):
    if probability < 0.30:
        return "LOW"
    elif probability < 0.60:
        return "MEDIUM"
    else:
        return "HIGH"
```

```python
student_index = 0

risk_probability = dropout_probabilities[student_index]
risk_level = get_risk_level(risk_probability)

print(f"Dropout risk: {risk_probability * 100:.1f}%")
print(f"Risk level: {risk_level}")
```

```python
def predict_student_risk(student_data):
    student_df = pd.DataFrame(
        [student_data],
        columns=early_features
    )

    probability = rf_model.predict_proba(student_df)[0][1]
    risk_level = get_risk_level(probability)

    return probability, risk_level
```
